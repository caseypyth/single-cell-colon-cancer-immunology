```python

# ۴. نصب Scanpy (فقط در صورت نیاز)

!pip install scanpy


# ۵. وارد کردن Scanpy به محیط کار

import scanpy as sc


# ۶. بارگذاری دیتاست PBMC-3k

# چون این یک Notebook تازه است، حافظه‌ی Notebook قبلی را نداریم.
adata = sc.datasets.pbmc3k()
```

```python
genes = ["CD3D", "MS4A1", "NKG7"]

for gene in genes:
    print(gene)
```

```python
cell_count = 0

while cell_count < 3:
    print("سلول بررسی شد.")
    cell_count = cell_count + 1
```

```python
test_genes = ["CD3D", "MT-ND1", "NKG7"]

for gene in test_genes:
    if gene.startswith("MT-"):
        print(gene, "--> Mitochondrial")
    else:
        print(gene, "--> Normal")
```

```python
# متغیری برای نگهداری تعداد ژن‌های میتوکندریایی
mt_count = 0

# حلقه روی تمام اسامی ژن‌ها
for gene in adata.var_names:
    if gene.startswith("MT-"):
        mt_count = mt_count + 1

print("تعداد ژن‌های میتوکندریایی پیدا شده:")
print(mt_count)
```
