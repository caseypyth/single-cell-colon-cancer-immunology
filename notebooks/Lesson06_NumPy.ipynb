```python
# ۵. یک آرایه‌ی ساختگی کوچک برای تمرین NumPy (۳ سلول × ۳ ژن فرضی)
arr = np.array([
    [5, 0, 12],
    [3, 8, 0],
    [0, 2, 7]
])

arr
```

```python
# ۶. نوع Object و ابعاد آرایه
type(arr)
arr.shape
```

```python
# ۷. ایندکس‌گذاری روی آرایه‌ی ساختگی
arr[0]        # سطر اول
arr[0, 0]     # یک خانه‌ی مشخص
arr[:, 0]     # ستون اول
```

```python
# ۸. عملیات ریاضی روی کل آرایه (بدون for دستی)
arr.sum()
arr.sum(axis=0)
arr.sum(axis=1)
arr.mean()
```
