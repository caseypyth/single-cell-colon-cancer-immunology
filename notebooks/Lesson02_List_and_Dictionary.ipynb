```python
genes = ["CD3D", "MS4A1", "NKG7"]

# اضافه کردن عضو جدید
genes.append("LYZ")
print("بعد از append:", genes)

# حذف عضو
genes.remove("MS4A1")
print("بعد از remove:", genes)

# تعداد اعضا
print("تعداد ژن‌ها:", len(genes))

# مرتب‌سازی
genes.sort()
print("بعد از sort:", genes)
```
