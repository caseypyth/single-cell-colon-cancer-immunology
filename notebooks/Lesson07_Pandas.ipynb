```python
# ===== درس ۷: Pandas پایه =====

# ۱. اتصال Google Drive
from google.colab import drive
drive.mount('/content/drive')

# ۲. تعریف مسیرهای پروژه با pathlib
from pathlib import Path

PROJECT = Path("/content/drive/MyDrive/scRNA_project")
LESSON = PROJECT / "lesson07"

CODE = LESSON / "code"
NOTES = LESSON / "notes"
OUTPUT = LESSON / "output"

# ۳. ساخت پوشه‌های همین درس
for folder in [CODE, NOTES, OUTPUT]:
    folder.mkdir(parents=True, exist_ok=True)
    print("آماده شد:", folder)

# ۴. نصب و بارگذاری Scanpy + NumPy + Pandas + دیتاست PBMC-3k
!pip install scanpy -q
import scanpy as sc
import numpy as np
import pandas as pd

adata = sc.datasets.pbmc3k()
```

```python
# ۵. DataFrame ساختگی برای تمرین
toy_data = {
    "gene": ["CD3D", "MS4A1", "NKG7"],
    "mean_expression": [5.2, 0.1, 3.4]
}
toy_df = pd.DataFrame(toy_data)
print(toy_df)

# ۶. فیلتر: فقط ژن‌هایی که میانگین بیانشون از ۱ بیشتره
print(toy_df[toy_df["mean_expression"] > 1])

# ۷. خلاصه‌ی آماری
print(toy_df.describe())
```

```python
# ۸. ماتریس بولین: کدوم خونه‌ها از adata.X غیرصفرن
is_expressed = adata.X > 0

# ۹. جمع هر سطر و صاف‌کردن به آرایه‌ی یک‌بعدی
gene_counts_per_cell = np.asarray(is_expressed.sum(axis=1)).flatten()

# ۱۰. اضافه‌کردن ستون جدید به adata.obs
adata.obs["n_genes"] = gene_counts_per_cell
print(adata.obs.head())
```

```python
# ۱۱. میانگین، خلاصه‌ی آماری و فیلتر روی داده‌ی واقعی
print(adata.obs["n_genes"].mean())
print(adata.obs["n_genes"].describe())
print(adata.obs[adata.obs["n_genes"] > 500].head())
```
